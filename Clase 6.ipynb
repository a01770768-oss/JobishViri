```python
import numpy as np
import pandas as pd
import seaborn as sns
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df.groupby('sex')[['survived']].mean()
```

```python
ndf = df.groupby(['sex', 'class'])['survived'].aggregate('mean').unstack()
ndf.index.name = None
ndf
```

```python
ndf = df.pivot_table('survived',
               index='sex',
               columns='class'
               )
ndf
```

```python
c = [1, 2, 2, 4, 5, 5, 5, 8, 9, 9, 9, 9]
```

```python
ccut = pd.cut(c, 3) #0.991 to  10 three bins
ccut
```

```python
ccut.value_counts()
```

```python
ccut2=pd.cut(c,
             [0, 5, 10]
)
ccut2
```

```python
cqcut = pd.qcut(c,3)
cqcut
```

```python
cqcut.value_counts()
```

Multi-level pivot tables



```python
age = pd.cut(df['age'],
             [0, 18, 80] #specify the bin edges
             )
age
```

```python
age[0]
```

```python
type[age]
```

```python
age.value_counts()
```

```python
df.pivot_table('survived',
               ['sex', age], #index columns
               'class' #columns
               )
```

```python
fare = pd.qcut(df['fare'], 2)
fare
```

```python
fare.value_counts()
```

```python
df.pivot_table('survived',
               ['sex', age],
               [fare, 'class']
               )
```

Extra Pivot Tables

```python
df.pivot_table(index = 'sex',
               columns = 'class',
               aggfunc = {'survived':sum,
                          'fare':'mean'
                          }
               )
```

```python
df.pivot_table('survived',
               index='sex',
               columns='class',
               margins=True,
               margins_name = 'Total'
               )
```
