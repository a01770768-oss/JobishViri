```python
import pandas as pd
import numpy as np
```

```python
x = {'Company': ['Ford', 'Ford', 'VW', 'Audi', 'Audi', 'Ford'],
     'Price': [2, 3, 1, 2, 3, 1],
     'Weight': [4, 6, 2, 2, 4, None],
     'Length': [6, 7, 3, 6, 8, np.nan],
     'Made': ['China', 'Mexico', 'Mexico', 'Correa', 'UK', pd.NaT]
    }
df = pd.DataFrame(data = x)
df
```

```python
pd.pivot_table(df[['Company', 'Price', 'Weight', 'Length']],
               index='Company'
              )
```

```python
pd.pivot_table(df,
               index='Company',
               values = 'Price'
              )
```

```python
x = pd.pivot_table(df,
               index='Company',
               values='Price', # will do the average of each a, b, c, d
               aggfunc = 'sum',
               #aggfunc =  ['sum', 'mean', 'count'],
               margins = True
              )#.head()
```

```python
x
```

```python
x.index.name = None
x
```

```python
df.groupby(['Company', 'Made'])[['Price']].sum()
```

```python
df['Price'].mean()
```

```python
pd.pivot_table(df,
               index='Company',
               values=['Price','Weight'], # will do the average of each a, b, c, d
               aggfunc = 'sum',
               margins = True,
               columns = 'Made'
              )#.head()
```

```python
x = {'Company': ['Ford', 'Ford', 'VW', 'Audi', 'Audi', 'Ford'],
     'Price': [2, 3, 1, 2, 3, 1],
     'Weight': [4, 6, 2, 2, 4, 3],
     'Length': [6, 9, 3, 6, 9, 3],
     'Made': ['China', 'Mexico', 'Mexico', 'China', 'China', 'China']
    }
df = pd.DataFrame(data = x)
df
```

```python
df
```

```python
pd.pivot_table(df,
               index='Company',
               values=['Price','Weight'], # will do the average of each a, b, c, d
               aggfunc={'Price': np.mean, 'Weight': np.sum},
               margins = True,
               margins_name='|*|Grand Total|*|',
               columns = 'Made'
              )#.head()
```
