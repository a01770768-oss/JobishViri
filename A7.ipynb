```python
import pandas as pd


df = pd.read_csv("antropometria-dataset-1.csv")

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(df["sistol"], bins=15)
plt.title("Sistol")
plt.xlabel("Sistol")
plt.ylabel("Frecuencia")

plt.subplot(1,2,2)
plt.hist(df["diastol"], bins=15)
plt.title("Diastol")
plt.xlabel("Diastol")
plt.ylabel("Frecuencia")

plt.show()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(df["peso"], bins=15)
plt.title("Peso")
plt.xlabel("Peso")
plt.ylabel("Frecuencia")

plt.subplot(1,2,2)
plt.hist(df["cintura"], bins=15)
plt.title("Cintura")
plt.xlabel("Cintura")
plt.ylabel("Frecuencia")

plt.show()
```

```python
plt.scatter(df["peso"], df["cintura"])

plt.title("Peso vs Cintura")
plt.xlabel("Peso")
plt.ylabel("Cintura")

plt.show()
```

```python
df[["peso", "cintura"]].corr()
```

```python
sexo1 = df[df["sexo"] == 1]
sexo2 = df[df["sexo"] == 2]
```

```python
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.hist(sexo1["sistol"], bins=15)
plt.title("Sistol - Sexo 1")

plt.subplot(2,2,2)
plt.hist(sexo1["diastol"], bins=15)
plt.title("Diastol - Sexo 1")

plt.subplot(2,2,3)
plt.hist(sexo1["peso"], bins=15)
plt.title("Peso - Sexo 1")

plt.subplot(2,2,4)
plt.hist(sexo1["cintura"], bins=15)
plt.title("Cintura - Sexo 1")

plt.show()
```

```python
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.hist(sexo2["sistol"], bins=15)
plt.title("Sistol - Sexo 2")

plt.subplot(2,2,2)
plt.hist(sexo2["diastol"], bins=15)
plt.title("Diastol - Sexo 2")

plt.subplot(2,2,3)
plt.hist(sexo2["peso"], bins=15)
plt.title("Peso - Sexo 2")

plt.subplot(2,2,4)
plt.hist(sexo2["cintura"], bins=15)
plt.title("Cintura - Sexo 2")

plt.show()
```

```python
plt.scatter(sexo1["peso"], sexo1["cintura"])

plt.title("Peso vs Cintura - Sexo 1")
plt.xlabel("Peso")
plt.ylabel("Cintura")

plt.show()
```

```python
plt.scatter(sexo2["peso"], sexo2["cintura"])

plt.title("Peso vs Cintura - Sexo 2")
plt.xlabel("Peso")
plt.ylabel("Cintura")

plt.show()
```
